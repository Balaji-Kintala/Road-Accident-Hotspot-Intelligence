```python
# ==========================================
# PROJECT 2 - ROAD ACCIDENT HOTSPOT INTELLIGENCE
# Stage 1: Data Loading & Initial Inspection
# ==========================================

import pandas as pd
import numpy as np

# ------------------------------------------
# 1. Load datasets
# ------------------------------------------

road_classification = pd.read_csv(
    "NCRB_ADSI_2023_Table_1A.7_2.csv"
)

time_occurrence = pd.read_csv(
    "NCRB_ADSI_2023_Table_1A.6_Road_Accidents_Time_of_Occurrence.csv"
)


# ------------------------------------------
# 2. Check dataset shapes
# ------------------------------------------

print("ROAD CLASSIFICATION DATASET")
print("Shape:", road_classification.shape)

print("\nTIME OF OCCURRENCE DATASET")
print("Shape:", time_occurrence.shape)


# ------------------------------------------
# 3. Display column names
# ------------------------------------------

print("\nROAD CLASSIFICATION COLUMNS")
print(road_classification.columns.tolist())

print("\nTIME OF OCCURRENCE COLUMNS")
print(time_occurrence.columns.tolist())


# ------------------------------------------
# 4. Display first 5 rows
# ------------------------------------------

print("\nROAD CLASSIFICATION - FIRST 5 ROWS")
display(road_classification.head())

print("\nTIME OF OCCURRENCE - FIRST 5 ROWS")
display(time_occurrence.head())


# ------------------------------------------
# 5. Display last 5 rows
# ------------------------------------------

print("\nROAD CLASSIFICATION - LAST 5 ROWS")
display(road_classification.tail())

print("\nTIME OF OCCURRENCE - LAST 5 ROWS")
display(time_occurrence.tail())
```

```python
# ==========================================
# PROJECT 2 - ROAD ACCIDENT HOTSPOT INTELLIGENCE
# Stage 2: Data Quality Check
# ==========================================

# ------------------------------------------
# 1. Check data types
# ------------------------------------------

print("ROAD CLASSIFICATION - DATA TYPES")
print(road_classification.dtypes)

print("\nTIME OF OCCURRENCE - DATA TYPES")
print(time_occurrence.dtypes)


# ------------------------------------------
# 2. Check missing values
# ------------------------------------------

print("\nROAD CLASSIFICATION - MISSING VALUES")
print(road_classification.isnull().sum())

print("\nTIME OF OCCURRENCE - MISSING VALUES")
print(time_occurrence.isnull().sum())


# ------------------------------------------
# 3. Check duplicate rows
# ------------------------------------------

print("\nROAD CLASSIFICATION - DUPLICATE ROWS")
print(road_classification.duplicated().sum())

print("\nTIME OF OCCURRENCE - DUPLICATE ROWS")
print(time_occurrence.duplicated().sum())


# ------------------------------------------
# 4. Check unique geography types
# ------------------------------------------

print("\nTIME OF OCCURRENCE - GEOGRAPHY TYPES")
print(time_occurrence["Geography Type"].value_counts())


# ------------------------------------------
# 5. Check aggregate / total rows
# ------------------------------------------

print("\nROAD CLASSIFICATION - TOTAL / AGGREGATE ROWS")

display(
    road_classification[
        road_classification["State/UT"]
        .astype(str)
        .str.contains("TOTAL", case=False, na=False)
    ]
)


print("\nTIME OF OCCURRENCE - TOTAL / AGGREGATE ROWS")

display(
    time_occurrence[
        time_occurrence["Geography"]
        .astype(str)
        .str.contains("TOTAL", case=False, na=False)
    ]
)


# ------------------------------------------
# 6. Check numerical summary
# ------------------------------------------

print("\nROAD CLASSIFICATION - NUMERICAL SUMMARY")
display(road_classification.describe())


print("\nTIME OF OCCURRENCE - NUMERICAL SUMMARY")
display(time_occurrence.describe())
```

```python
# ==========================================
# Stage 3: Data Consistency Validation
# ==========================================

# ------------------------------------------
# 1. Validate Road Classification totals
# ------------------------------------------

road_classification["Calculated Cases"] = (
    road_classification["National Highways - Cases"]
    + road_classification["State Highways - Cases"]
    + road_classification["Expressways - Cases"]
    + road_classification["Other Roads - Cases"]
)

road_classification["Calculated Injured"] = (
    road_classification["National Highways - Injured"]
    + road_classification["State Highways - Injured"]
    + road_classification["Expressways - Injured"]
    + road_classification["Other Roads - Injured"]
)

road_classification["Calculated Died"] = (
    road_classification["National Highways - Died"]
    + road_classification["State Highways - Died"]
    + road_classification["Expressways - Died"]
    + road_classification["Other Roads - Died"]
)


# ------------------------------------------
# 2. Compare calculated totals with reported
# ------------------------------------------

print("CASES TOTAL MATCH:")
print(
    (road_classification["Calculated Cases"]
     == road_classification["Total - Cases"]).all()
)

print("\nINJURED TOTAL MATCH:")
print(
    (road_classification["Calculated Injured"]
     == road_classification["Total - Injured"]).all()
)

print("\nDIED TOTAL MATCH:")
print(
    (road_classification["Calculated Died"]
     == road_classification["Total - Died"]).all()
)


# ------------------------------------------
# 3. Show any mismatches
# ------------------------------------------

print("\nCASE TOTAL MISMATCHES")
display(
    road_classification[
        road_classification["Calculated Cases"]
        != road_classification["Total - Cases"]
    ]
)

print("\nINJURED TOTAL MISMATCHES")
display(
    road_classification[
        road_classification["Calculated Injured"]
        != road_classification["Total - Injured"]
    ]
)

print("\nDEATH TOTAL MISMATCHES")
display(
    road_classification[
        road_classification["Calculated Died"]
        != road_classification["Total - Died"]
    ]
)


# ------------------------------------------
# 4. Validate Time of Occurrence totals
# ------------------------------------------

time_columns = [
    "0000-0300 (Night)",
    "0300-0600 (Night)",
    "0600-0900 (Day)",
    "0900-1200 (Day)",
    "1200-1500 (Day)",
    "1500-1800 (Day)",
    "1800-2100 (Night)",
    "2100-2400 (Night)"
]

time_occurrence["Calculated Total"] = (
    time_occurrence[time_columns].sum(axis=1)
)


print("\nTIME OF OCCURRENCE TOTAL MATCH:")
print(
    (time_occurrence["Calculated Total"]
     == time_occurrence["Total"]).all()
)


# ------------------------------------------
# 5. Show any time-total mismatches
# ------------------------------------------

print("\nTIME TOTAL MISMATCHES")

display(
    time_occurrence[
        time_occurrence["Calculated Total"]
        != time_occurrence["Total"]
    ]
)
```

```python
# ==========================================
# PROJECT 2 - ROAD ACCIDENT HOTSPOT INTELLIGENCE
# Stage 4: Data Cleaning
# ==========================================

# ------------------------------------------
# 1. Remove aggregate rows from
#    Road Classification dataset
# ------------------------------------------

road_classification = road_classification[
    ~road_classification["State/UT"]
    .astype(str)
    .str.contains("Total", case=False, na=False)
].copy()


# ------------------------------------------
# 2. Remove temporary validation columns
# ------------------------------------------

road_classification = road_classification.drop(
    columns=[
        "Calculated Cases",
        "Calculated Injured",
        "Calculated Died"
    ]
)


# ------------------------------------------
# 3. Remove aggregate rows from
#    Time of Occurrence dataset
# ------------------------------------------

time_occurrence = time_occurrence[
    ~time_occurrence["Geography"]
    .astype(str)
    .str.contains("Total", case=False, na=False)
].copy()


# ------------------------------------------
# 4. Remove temporary validation column
# ------------------------------------------

time_occurrence = time_occurrence.drop(
    columns=["Calculated Total"]
)


# ------------------------------------------
# 5. Reset indexes
# ------------------------------------------

road_classification.reset_index(drop=True, inplace=True)
time_occurrence.reset_index(drop=True, inplace=True)


# ------------------------------------------
# 6. Check cleaned shapes
# ------------------------------------------

print("ROAD CLASSIFICATION - CLEANED SHAPE:")
print(road_classification.shape)

print("\nTIME OF OCCURRENCE - CLEANED SHAPE:")
print(time_occurrence.shape)


# ------------------------------------------
# 7. Verify no aggregate rows remain
# ------------------------------------------

print("\nROAD CLASSIFICATION - TOTAL ROWS REMAINING:")
print(
    road_classification["State/UT"]
    .astype(str)
    .str.contains("Total", case=False, na=False)
    .sum()
)

print("\nTIME OF OCCURRENCE - TOTAL ROWS REMAINING:")
print(
    time_occurrence["Geography"]
    .astype(str)
    .str.contains("Total", case=False, na=False)
    .sum()
)


# ------------------------------------------
# 8. Final missing-value check
# ------------------------------------------

print("\nROAD CLASSIFICATION - TOTAL MISSING VALUES:")
print(road_classification.isnull().sum().sum())

print("\nTIME OF OCCURRENCE - TOTAL MISSING VALUES:")
print(time_occurrence.isnull().sum().sum())


# ------------------------------------------
# 9. Final duplicate check
# ------------------------------------------

print("\nROAD CLASSIFICATION - DUPLICATES:")
print(road_classification.duplicated().sum())

print("\nTIME OF OCCURRENCE - DUPLICATES:")
print(time_occurrence.duplicated().sum())
```

```python
# ==========================================
# PROJECT 2 - ROAD ACCIDENT HOTSPOT INTELLIGENCE
# Stage 5: Feature Engineering
# ==========================================


# ==================================================
# PART A: ROAD CLASSIFICATION DATASET
# ==================================================

# ------------------------------------------
# 1. Accident Share (%)
# ------------------------------------------

road_classification["Accident Share (%)"] = (
    road_classification["Total - Cases"]
    / road_classification["Total - Cases"].sum()
) * 100


# ------------------------------------------
# 2. Injury Rate (%)
# ------------------------------------------

road_classification["Injury Rate (%)"] = (
    road_classification["Total - Injured"]
    / road_classification["Total - Cases"]
) * 100


# ------------------------------------------
# 3. Fatality Rate (%)
# ------------------------------------------

road_classification["Fatality Rate (%)"] = (
    road_classification["Total - Died"]
    / road_classification["Total - Cases"]
) * 100


# ==================================================
# PART B: TIME OF OCCURRENCE DATASET
# ==================================================

# ------------------------------------------
# 4. Create Day Accident Total
# ------------------------------------------

day_columns = [
    "0600-0900 (Day)",
    "0900-1200 (Day)",
    "1200-1500 (Day)",
    "1500-1800 (Day)"
]

time_occurrence["Day Accidents"] = (
    time_occurrence[day_columns].sum(axis=1)
)


# ------------------------------------------
# 5. Create Night Accident Total
# ------------------------------------------

night_columns = [
    "0000-0300 (Night)",
    "0300-0600 (Night)",
    "1800-2100 (Night)",
    "2100-2400 (Night)"
]

time_occurrence["Night Accidents"] = (
    time_occurrence[night_columns].sum(axis=1)
)


# ------------------------------------------
# 6. Day Accident Share (%)
# ------------------------------------------

time_occurrence["Day Accident Share (%)"] = (
    time_occurrence["Day Accidents"]
    / time_occurrence["Total"]
) * 100


# ------------------------------------------
# 7. Night Accident Share (%)
# ------------------------------------------

time_occurrence["Night Accident Share (%)"] = (
    time_occurrence["Night Accidents"]
    / time_occurrence["Total"]
) * 100


# ------------------------------------------
# 8. Identify Peak Time Period
# ------------------------------------------

time_occurrence["Peak Time Period"] = (
    time_occurrence[time_columns]
    .idxmax(axis=1)
)


# ------------------------------------------
# 9. Get Peak Time Accident Count
# ------------------------------------------

time_occurrence["Peak Time Accidents"] = (
    time_occurrence[time_columns]
    .max(axis=1)
)


# ==================================================
# PART C: CHECK THE NEW FEATURES
# ==================================================

print("ROAD CLASSIFICATION - NEW FEATURES")

display(
    road_classification[
        [
            "State/UT",
            "Total - Cases",
            "Total - Injured",
            "Total - Died",
            "Accident Share (%)",
            "Injury Rate (%)",
            "Fatality Rate (%)"
        ]
    ].head()
)


print("\nTIME OF OCCURRENCE - NEW FEATURES")

display(
    time_occurrence[
        [
            "Geography Type",
            "Geography",
            "Day Accidents",
            "Night Accidents",
            "Day Accident Share (%)",
            "Night Accident Share (%)",
            "Peak Time Period",
            "Peak Time Accidents"
        ]
    ].head()
)


# ==================================================
# PART D: VALIDATE FEATURE ENGINEERING
# ==================================================

print("\nDAY + NIGHT ACCIDENTS SHOULD EQUAL TOTAL")

print(
    (
        time_occurrence["Day Accidents"]
        + time_occurrence["Night Accidents"]
        == time_occurrence["Total"]
    ).all()
)
```

```python
# ==========================================
# Stage 6: Export Cleaned Analytical Data
# ==========================================

road_classification.to_csv(
    "road_classification_cleaned.csv",
    index=False
)

time_occurrence.to_csv(
    "time_occurrence_cleaned.csv",
    index=False
)

print("Files exported successfully.")

print("\nRoad Classification:")
print(road_classification.shape)

print("\nTime of Occurrence:")
print(time_occurrence.shape)
```
